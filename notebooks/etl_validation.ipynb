ROW COUNT VALIDATION

Customers Validation

```python
%sql
SELECT 'bronze_customers' AS table_name,
       COUNT(*) AS row_count
FROM `retail-sales-proj-t9`.bronze.customers_raw

UNION ALL

SELECT 'silver_customers',
       COUNT(*)
FROM `retail-sales-proj-t9`.silver.customers_silver

UNION ALL

SELECT 'gold_customers',
       COUNT(*)
FROM `retail-sales-proj-t9`.gold.dim_customer;
```

Products Validation

```python
%sql
SELECT 'bronze_products', COUNT(*) FROM `retail-sales-proj-t9`.bronze.products_raw

UNION ALL

SELECT 'silver_products', COUNT(*) FROM `retail-sales-proj-t9`.silver.products_silver

UNION ALL

SELECT 'gold_products', COUNT(*) FROM `retail-sales-proj-t9`.gold.dim_product;
```

Sales Validation

```python
%sql
SELECT 'bronze_sales', COUNT(*) FROM `retail-sales-proj-t9`.bronze.sales_raw

UNION ALL

SELECT 'silver_sales', COUNT(*) FROM `retail-sales-proj-t9`.silver.sales_silver

UNION ALL

SELECT 'gold_sales', COUNT(*) FROM `retail-sales-proj-t9`.gold.fact_sales;
```

COLUMN MAPPING VALIDATION

```python
%sql
SELECT

    b.CustomerID,
    b.CustomerName,

    s.customer_id,
    s.customer_name

FROM `retail-sales-proj-t9`.bronze.customers_raw b

JOIN `retail-sales-proj-t9`.silver.customers_silver s

ON b.CustomerID = s.customer_id

LIMIT 10;
```

DATA TYPE VALIDATION

```python
%sql
DESCRIBE `retail-sales-proj-t9`.silver.customers_silver;
```

TRIM VALIDATION

```python
%sql
SELECT *
FROM `retail-sales-proj-t9`.silver.customers_silver
WHERE customer_name != TRIM(customer_name);
```

LOWERCASE EMAIL VALIDATION

```python
%sql
SELECT *
FROM `retail-sales-proj-t9`.silver.customers_silver
WHERE email != LOWER(email);
```

PROPER CASE VALIDATION

```python
%sql
SELECT *
FROM `retail-sales-proj-t9`.silver.products_silver
WHERE product_name != INITCAP(product_name);
```

DERIVED AMOUNT VALIDATION

```python
%sql
SELECT
    quantity,
    unit_price,
    total_amount,
    (quantity * unit_price) AS expected_total

FROM `retail-sales-proj-t9`.gold.fact_sales

WHERE total_amount != (quantity * unit_price);
```

DATE FORMAT VALIDATION

```python
%sql
SELECT *
FROM `retail-sales-proj-t9`.silver.sales_silver
WHERE transaction_date IS NULL;
```

DUPLICATE VALIDATION

```python
%sql
SELECT
    customer_id,
    COUNT(*)

FROM `retail-sales-proj-t9`.silver.customers_silver

GROUP BY customer_id

HAVING COUNT(*) > 1;
```

NULL VALIDATION

```python
%sql
SELECT *
FROM `retail-sales-proj-t9`.gold.dim_customer
WHERE customer_name IS NULL;
```

INVALID DATA VALIDATION

```python
%sql
SELECT *
FROM `retail-sales-proj-t9`.silver.sales_silver
WHERE quantity <= 0;
```

REFERENTIAL INTEGRITY VALIDATION

```python
%sql
SELECT *
FROM `retail-sales-proj-t9`.gold.fact_sales f

LEFT JOIN `retail-sales-proj-t9`.gold.dim_customer c
ON f.customer_id = c.customer_id

WHERE c.customer_id IS NULL;
```

ACTIVE VS INACTIVE RECORDS

```python
%sql
SELECT
    customer_id,
    customer_name,
    city,
    is_active,
    start_date,
    end_date

FROM `retail-sales-proj-t9`.gold.dim_customer

ORDER BY customer_id;
```

ONLY ONE ACTIVE RECORD

```python
%sql
SELECT
    customer_id,
    COUNT(*)

FROM `retail-sales-proj-t9`.gold.dim_customer

WHERE is_active = 1

GROUP BY customer_id

HAVING COUNT(*) > 1;
```

Validation of Incremental Load

```python
%sql
SELECT *
FROM `retail-sales-proj-t9`.gold.dim_customer
WHERE customer_id = 1
ORDER BY start_date;
```
