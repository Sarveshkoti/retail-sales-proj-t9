```python
%sql

USE CATALOG `retail-sales-proj-t9`;

CREATE SCHEMA IF NOT EXISTS bronze;

------------------------------------------------
-- CUSTOMERS
------------------------------------------------

CREATE OR REPLACE TABLE bronze.customers_raw
USING DELTA
AS

SELECT *
FROM read_files(

    's3://retail-sales-proj-t9/raw/customers/',

    format => 'csv',
    header => true,
    inferSchema => true

);

------------------------------------------------
-- PRODUCTS
------------------------------------------------

CREATE OR REPLACE TABLE bronze.products_raw
USING DELTA
AS

SELECT *
FROM read_files(

    's3://retail-sales-proj-t9/raw/products/',

    format => 'csv',
    header => true,
    inferSchema => true

);

------------------------------------------------
-- STORES
------------------------------------------------

CREATE OR REPLACE TABLE bronze.stores_raw
USING DELTA
AS

SELECT *
FROM read_files(

    's3://retail-sales-proj-t9/raw/stores/',

    format => 'csv',
    header => true,
    inferSchema => true

);

------------------------------------------------
-- SALES
------------------------------------------------

CREATE OR REPLACE TABLE bronze.sales_raw
USING DELTA
AS

SELECT *
FROM read_files(

    's3://retail-sales-proj-t9/raw/sales/',

    format => 'csv',
    header => true,
    inferSchema => true

);
```

```python
%sql
SELECT * FROM bronze.customers_raw;
```

```python
%sql
SELECT 'customers' AS table_name, COUNT(*) AS row_count FROM bronze.customers_raw
UNION ALL
SELECT 'products',  COUNT(*) FROM bronze.products_raw
UNION ALL
SELECT 'stores',    COUNT(*) FROM bronze.stores_raw
UNION ALL
SELECT 'sales',     COUNT(*) FROM bronze.sales_raw;
```

```python
%sql
SELECT * FROM bronze.sales_raw;
```
